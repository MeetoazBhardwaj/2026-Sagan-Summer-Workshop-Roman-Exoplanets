```python
# Here's an example for how to generate a SPISEA isochrone:

# Define isochrone parameters
logAge = 10 # Age in log(years)
AKs = 0.4 # extinction in mags
dist = 8000 # distance in parsec
metal = 0.0 # metallicity
iso_dir = f'{this_dir}/isochrones/'

# Define evolution/atmosphere models and extinction law (optional)
evo_model = evolution.MISTv1()
atm_func = atmospheres.get_merged_atmosphere
red_law = reddening.RedLawHosek18b()

# Also specify filters for synthetic photometry (optional)
filt_list = ['roman,wfi,f062', 'roman,wfi,f087', 'roman,wfi,f106',
             'roman,wfi,f129', 'roman,wfi,f158', 'roman,wfi,f146',
             'roman,wfi,f184', 'roman,wfi,f213']

# Create (or read in) the isochrone
# Be sure to set your mag_sys to 'AB' as is standard for Roman and not SPISEA's default.
iso = syn.IsochronePhot(logAge = logAge,
                  AKs = AKs,
                  distance = dist,
                  metallicity = metal,
                  iso_dir = iso_dir,
                  evo_model = evo_model,
                  atm_func = atm_func,
                  red_law=red_law,
                  filters=filt_list,
                  mag_sys='AB'
                  )
```

# Dark Lenses

Some microlensing events have very little flux contamination from a luminous lens (i.e. the source flux fraction is close to b_sff = 1). In these cases, we have several other constraining properties on the source, including its parallax and proper motion. This can be useful in assessing the properties of the source and decoupling them from properties of the lens.
